```python
# Define a function to interpolate heat flux grid at a given time
def interpolate_heat_flux_grid(heat_flux_grid, enhancements_grid, list_masses, 
                               interp_enhancements=None):
    """
    Interpolate a heat flux grid using linear fits for each mass.
    
    Parameters:
    -----------
    heat_flux_grid : list of lists
        Heat flux grid where each row corresponds to a mass
    enhancements_grid : list of lists
        Enhancement factors for each mass
    list_masses : list
        List of planetary masses
    interp_enhancements : array, optional
        Enhancement values at which to interpolate (default: 1 to 10000 with 10000 points)
    
    Returns:
    --------
    heat_flux_grid_interpolated : list
        Interpolated heat flux grid
    linear_fits : list
        Linear fit objects for each mass
    interp_enhancements : array
        Enhancement values used for interpolation
    """
    if interp_enhancements is None:
        interp_enhancements = np.linspace(1, 10000, 10000)
    
    heat_flux_grid_interpolated = []
    heat_flux_floor = 1e-4 # Set a floor for heat flux values to prevent negative values
    linear_fits = []
    
    for i, mass in enumerate(list_masses):
        # Remove NaN values for fitting
        valid_indices = ~np.isnan(heat_flux_grid[i])
        valid_enhancements = np.array(enhancements_grid[i])[valid_indices]
        valid_heat_fluxes = np.array(heat_flux_grid[i])[valid_indices]
        
        # Fit a linear function to this mass's data in log space
        if len(valid_heat_fluxes) > 1:
            coeffs = np.polyfit(np.log10(valid_enhancements), np.log10(valid_heat_fluxes), 1)
            linear_fit = np.poly1d(coeffs)
            linear_fits.append(linear_fit)
            
            # Interpolate/extrapolate across the enhancement range
            interp_heat_fluxes = linear_fit(np.log10(interp_enhancements))
            interp_heat_fluxes = 10**interp_heat_fluxes

            # Apply floor to prevent negative values
            interp_heat_fluxes = np.maximum(interp_heat_fluxes, heat_flux_floor)
            heat_flux_grid_interpolated.append(interp_heat_fluxes)
        else:
            print(f"Warning: mass {mass} has insufficient valid data points for fitting")
            linear_fits.append(None)
            heat_flux_grid_interpolated.append(None)
    
    return heat_flux_grid_interpolated, linear_fits, interp_enhancements
```

```python
# Helper function to extract heat flux at a given enhancement level
def get_heat_flux_vs_mass(heat_flux_grid_interpolated, interp_enhancements, enhancement_value):
    """
    Extract heat flux as a function of mass at a specific enhancement level.
    
    Parameters:
    -----------
    heat_flux_grid_interpolated : list
        Interpolated heat flux grid
    interp_enhancements : array
        Enhancement values used for interpolation
    enhancement_value : float
        Target enhancement level
    
    Returns:
    --------
    heat_flux_slice : array
        Heat flux values for each mass at the given enhancement level
    """
    # Find the index closest to the requested enhancement
    idx = np.argmin(np.abs(interp_enhancements - enhancement_value))
    
    # Extract heat flux at this index for all masses
    heat_flux_slice = np.array([heat_flux_grid_interpolated[i][idx] 
                                for i in range(len(heat_flux_grid_interpolated))])
    return heat_flux_slice
```

```python
# define a function to find the enhancement value where the heat flux crosses a certain value
def heat_flux_enhance_crossing(heat_flux_grid_interpolated, interp_enhancements, target_heat_flux):
    """
    Find the enhancement value where the heat flux crosses a target value.
    
    Parameters:
    -----------
    heat_flux_grid_interpolated : list
        Interpolated heat flux grid for a specific mass
    interp_enhancements : array
        Enhancement values corresponding to the interpolated heat flux grid
    target_heat_flux : float
        Target heat flux value to find the crossing point for
    
    Returns:
    --------
    crossing_enhancement : float or None
        Enhancement value where the heat flux crosses the target value, or None if no crossing is found
    """
    if heat_flux_grid_interpolated is None:
        return None
    
    # Find indices where the heat flux crosses the target value
    crossing_indices = np.where(np.diff(np.sign(heat_flux_grid_interpolated - target_heat_flux)))[0]
    
    if len(crossing_indices) > 0:
        # Take the first crossing point (or you could average multiple crossings if needed)
        idx = crossing_indices[0]
        # Linear interpolation to find the exact crossing enhancement
        x1, x2 = interp_enhancements[idx], interp_enhancements[idx + 1]
        y1, y2 = heat_flux_grid_interpolated[idx], heat_flux_grid_interpolated[idx + 1]
        
        if y2 != y1:  # Avoid division by zero
            crossing_enhancement = x1 + (target_heat_flux - y1) * (x2 - x1) / (y2 - y1)
            return crossing_enhancement
        else:
            return x1  # If both points have the same heat flux, return either enhancement value
    else:
        return None
```

```python
# and now one for the other direction: for a given enhancement, find the heat flux
def heat_flux_at_enhancement(heat_flux_grid_interpolated, interp_enhancements, target_enhancement):
    """
    Find the heat flux at a specific enhancement value.
    
    Parameters:
    -----------
    heat_flux_grid_interpolated : list
        Interpolated heat flux grid for a specific mass
    interp_enhancements : array
        Enhancement values corresponding to the interpolated heat flux grid
    target_enhancement : float
        Target enhancement value to for which to find the corresponding heat flux, in log units (dex [X/Fe])
    
    Returns:
    --------
    heat_flux_at_target : float or None
        Heat flux corresponding to the target enhancement, or None if target enhancement is out of bounds
    """
    if heat_flux_grid_interpolated is None:
        return None
    
    target_enhancement = 10 ** target_enhancement  # Convert from log units to linear scale
    
    # Check if target enhancement is within the range of interpolated enhancements
    if target_enhancement < interp_enhancements[0] or target_enhancement > interp_enhancements[-1]:
        return None
    
    # Find indices surrounding the target enhancement
    idx = np.searchsorted(interp_enhancements, target_enhancement) - 1
    
    if idx >= 0 and idx < len(interp_enhancements) - 1:
        x1, x2 = interp_enhancements[idx], interp_enhancements[idx + 1]
        y1, y2 = heat_flux_grid_interpolated[idx], heat_flux_grid_interpolated[idx + 1]
        
        if x2 != x1:  # Avoid division by zero
            heat_flux_at_target = y1 + (target_enhancement - x1) * (y2 - y1) / (x2 - x1)
            return heat_flux_at_target
        else:
            return y1  # If both points have the same enhancement, return either heat flux value
    else:
        return None
```
