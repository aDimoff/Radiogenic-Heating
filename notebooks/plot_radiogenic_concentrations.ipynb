```python
# read mass_sweep data from files within directory
data_dir = 'test_radiogenic_concentrations/'

heat_flux_files = !ls {data_dir}*heat_flux*
print(heat_flux_files[0])

#heat_flow_files = !ls {data_dir}*heat_flow*
#print(heat_flow_files[0])
```

```python
def extract_heat_fluxes(file):
    data = np.genfromtxt(file, unpack=True)
    time = data[0]
    conv_heat_flux = data[1]
    mag_heat_flux = data[2]
    total_heat_flux = conv_heat_flux + mag_heat_flux
    return time, conv_heat_flux, mag_heat_flux, total_heat_flux
```

```python
def extract_heat_fluxes_2(file):
    data = np.genfromtxt(file, unpack=True)
    time = data[0]
    conv_heat_flux = data[1]
    mag_heat_flux = data[2]
    surf_heat_flux = data[3]
    total_surface_heat_flux = surf_heat_flux + mag_heat_flux
    return time, conv_heat_flux, mag_heat_flux, surf_heat_flux, total_surface_heat_flux
```

```python
earth_time = 4.5 #Gyr

fig, (ax1, ax2, ax3, ax4) = plt.subplots(1, 4, figsize=(12, 3), dpi=200, sharey=True, sharex=True, constrained_layout=True)

# First, get Earth standard data
time_earth, conv_earth, mag_earth, total_earth = extract_heat_fluxes('test_radiogenic_concentrations/Earth_heat_fluxWm2.dat')

# Plot Earth standard on all subplots
for ax in [ax1, ax2, ax3, ax4]:
    ax.plot(time_earth, total_earth, label='Total', ls='-', color='black', lw=2)
    ax.plot(time_earth, conv_earth, label='Convective', ls='--', color='grey', lw=1.0, alpha=0.7)
    ax.plot(time_earth, mag_earth, label='Magmatic', ls=':', color='grey', lw=1.0, alpha=0.7)

# Store data for each isotope variation
U235_data = {'0p5': None, '1p5': None}
U238_data = {'0p5': None, '1p5': None}
Th_data = {'0p5': None, '1p5': None}
K_data = {'0p5': None, '1p5': None}

for file in heat_flux_files:
    # Skip the base Earth file (exact match only)
    if file.endswith('Earth_heat_fluxWm2.dat'):
        continue
        
    if 'U235' in file:
        if '0p5' in file:
            U235_data['0p5'] = extract_heat_fluxes(file)
        elif '1p5' in file:
            U235_data['1p5'] = extract_heat_fluxes(file)
    
    elif 'U238' in file:
        if '0p5' in file:
            U238_data['0p5'] = extract_heat_fluxes(file)
        elif '1p5' in file:
            U238_data['1p5'] = extract_heat_fluxes(file)
    
    elif 'Th' in file:
        if '0p5' in file:
            Th_data['0p5'] = extract_heat_fluxes(file)
        elif '1p5' in file:
            Th_data['1p5'] = extract_heat_fluxes(file)

    elif '_K' in file or 'K0' in file or 'K1' in file:
        if '0p5' in file:
            K_data['0p5'] = extract_heat_fluxes(file)
        elif '1p5' in file:
            K_data['1p5'] = extract_heat_fluxes(file)

# Interpolate and fill between for each isotope - including convective and magmatic envelopes
def fill_isotope_range(ax, data_dict, time_ref, color='lightblue'):
    """Interpolate 0.5x and 1.5x data onto reference time grid and fill between"""
    if data_dict['0p5'] is not None and data_dict['1p5'] is not None:
        time05, conv05, mag05, total05 = data_dict['0p5']
        time15, conv15, mag15, total15 = data_dict['1p5']
        
        # Interpolate onto Earth time grid
        conv05_interp = np.interp(time_ref, time05, conv05)
        conv15_interp = np.interp(time_ref, time15, conv15)
        mag05_interp = np.interp(time_ref, time05, mag05)
        mag15_interp = np.interp(time_ref, time15, mag15)
        total05_interp = np.interp(time_ref, time05, total05)
        total15_interp = np.interp(time_ref, time15, total15)
        
        # Fill between for total (most opaque)
        ax.fill_between(time_ref, total05_interp, total15_interp, 
                       color=color, alpha=0.35, label='±50% Total', zorder=1)
        
        # Fill between for convective (lighter)
        #ax.fill_between(time_ref, conv05_interp, conv15_interp, 
        #               color=color, alpha=0.2, linestyle='--', zorder=2)
        
        # Fill between for magmatic (lighter)
        #ax.fill_between(time_ref, mag05_interp, mag15_interp, 
        #               color=color, alpha=0.2, linestyle=':', zorder=2)

fill_isotope_range(ax1, U235_data, time_earth, color='steelblue')
fill_isotope_range(ax2, U238_data, time_earth, color='steelblue')
fill_isotope_range(ax3, Th_data, time_earth, color='steelblue')
fill_isotope_range(ax4, K_data, time_earth, color='steelblue')

# Add text for the isotope being plotted in each subplot
ax1.text(s=r'$^{235}$U $\pm \; 50\%$', x=0.70, y=0.90, transform=ax1.transAxes, ha='center', va='top', fontsize=20)
ax2.text(s=r'$^{238}$U $\pm \; 50\%$', x=0.70, y=0.90, transform=ax2.transAxes, ha='center', va='top', fontsize=20)
ax3.text(s=r'$^{232}$Th $\pm \; 50\%$', x=0.70, y=0.90, transform=ax3.transAxes, ha='center', va='top', fontsize=20)
ax4.text(s=r'$^{40}$K $\pm \; 50\%$', x=0.70, y=0.90, transform=ax4.transAxes, ha='center', va='top', fontsize=20)

# Add text for the half-life of each isotope in each subplot
ax1.text(s=r'$t_{1/2} = 0.704$ Gyr', x=0.75, y=0.75, transform=ax1.transAxes, ha='center', va='top', fontsize=10)
ax2.text(s=r'$t_{1/2} = 4.468$ Gyr', x=0.75, y=0.75, transform=ax2.transAxes, ha='center', va='top', fontsize=10)
ax3.text(s=r'$t_{1/2} = 14.05$ Gyr', x=0.75, y=0.75, transform=ax3.transAxes, ha='center', va='top', fontsize=10)
ax4.text(s=r'$t_{1/2} = 1.248$ Gyr', x=0.75, y=0.75, transform=ax4.transAxes, ha='center', va='top', fontsize=10)

ax1.set_ylim(0,0.20)
ax1.set_xlim(0.2, 10)
ax1.set_xscale('log')
ax1.set_ylim(1e-2,0.32); ax1.set_yscale('log')
ax1.set_xlabel('Time (Gyr)'); ax2.set_xlabel('Time (Gyr)'); ax3.set_xlabel('Time (Gyr)'); ax4.set_xlabel('Time (Gyr)')
ax1.set_ylabel('Heat Flux (W/m$^2$)')

# Add legend to first subplot only
ax1.legend(fontsize=7, loc='center right', framealpha=0.9)
```

```python
# what is the relative width of the error envelope at Earth time for each isotope?
def relative_width_at_earth_time(data_dict, time_ref):
    if data_dict['0p5'] is not None and data_dict['1p5'] is not None:
        time05, conv05, mag05, total05 = data_dict['0p5']
        time15, conv15, mag15, total15 = data_dict['1p5']
        
        # Interpolate onto Earth time grid
        total05_interp = np.interp(time_ref, time05, total05)
        total15_interp = np.interp(time_ref, time15, total15)
        
        # Calculate relative width
        relative_width = (total15_interp - total05_interp) / ((total15_interp + total05_interp) / 2)
        return relative_width
    else:
        return None
```

```python
relative_width_U235 = relative_width_at_earth_time(U235_data, earth_time)
relative_width_U238 = relative_width_at_earth_time(U238_data, earth_time)
relative_width_Th = relative_width_at_earth_time(Th_data, earth_time)
relative_width_K = relative_width_at_earth_time(K_data, earth_time)

print(f"Relative width of ±50% envelope at Earth time:")
print(f"U235: {relative_width_U235:.2f}")
print(f"U238: {relative_width_U238:.2f}")
print(f"Th: {relative_width_Th:.2f}")
print(f"K: {relative_width_K:.2f}")
```
