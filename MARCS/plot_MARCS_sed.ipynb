```python
# make a function to integrate the SED to get the bolometric flux
def integrate_SED(wavelength, flux):
    # convert wavelength from Angstroms to cm
    wavelength_cm = wavelength * 1e-8
    # convert flux from erg s^-1 cm^-2 A^-1 to erg s^-1 cm^-2 cm^-1
    flux_cm = flux * 1e8
    # integrate using the trapezoidal rule
    bolometric_flux = np.trapezoid(flux_cm, wavelength_cm)
    return bolometric_flux

wavs = np.arange(0, len(g_0_SED[::10]))

# compute the bolometric flux for the G0 model
F_flux_bol = integrate_SED(wavs, f_0_SED[::10])
G_flux_bol = integrate_SED(wavs, g_0_SED[::10]) 
K_flux_bol = integrate_SED(wavs, k_0_SED[::10]) 
M_flux_bol = integrate_SED(wavs, m_0_SED[::10]) 
# convert from erg s^-1 cm^-2 to W m^-2
F_flux_bol_Wm2 = F_flux_bol * 1e-3
G_flux_bol_Wm2 = G_flux_bol * 1e-3
K_flux_bol_Wm2 = K_flux_bol * 1e-3
M_flux_bol_Wm2 = M_flux_bol * 1e-3
print(f'Bolometric flux for F0 model: {F_flux_bol_Wm2:.2e} W m^-2')
print(f'Bolometric flux for G0 model: {G_flux_bol_Wm2:.2e} W m^-2')
print(f'Bolometric flux for K0 model: {K_flux_bol_Wm2:.2e} W m^-2')
print(f'Bolometric flux for M0 model: {M_flux_bol_Wm2:.2e} W m^-2')


max_F_flux_bol = integrate_SED(wavs, max_f_SED[::10]) * 1e-3
min_M_flux_bol = integrate_SED(wavs, min_m_SED[::10]) * 1e-3
print(f'Bolometric flux for max F model: {max_F_flux_bol:.2e} W m^-2')
print(f'Bolometric flux for min M model: {min_M_flux_bol:.2e} W m^-2')
```

```python
# compute luminosity from flux and distance
Earth_distance_cm = 1.496e13 # cm
Jupiter_distance_cm = 7.785e13 # cm

distances = np.linspace(0.01, 100, 1000) * Earth_distance_cm

# for the G stars, we can assume a stellar radius of 1 solar radius
solar_radius = 6.957e10 # cm
# f star radius are about 1.3 solar radii, m star radii are about 0.5 solar radii
f_star_radius = 1.3 * solar_radius
k_star_radius = 0.8 * solar_radius
m_star_radius = 0.5 * solar_radius

def compute_luminosity(Fbol, stellar_radius):
    L = 4 * np.pi * stellar_radius**2 * Fbol
    return L
```

```python
# compute earth-like luminosities
f_0_L = compute_luminosity(f_0_Fbol, f_star_radius)
f_m1_L = compute_luminosity(f_m1_Fbol, f_star_radius)
f_m2_L = compute_luminosity(f_m2_Fbol, f_star_radius)
f_m3_L = compute_luminosity(f_m3_Fbol, f_star_radius)

g_0_L = compute_luminosity(g_0_Fbol, solar_radius)
g_m1_L = compute_luminosity(g_m1_Fbol, solar_radius)
g_m2_L = compute_luminosity(g_m2_Fbol, solar_radius)
g_m3_L = compute_luminosity(g_m3_Fbol, solar_radius)

g_g1_L = compute_luminosity(g_g1_Fbol, solar_radius)
g_g2_L = compute_luminosity(g_g2_Fbol, solar_radius)
g_g3_L = compute_luminosity(g_g3_Fbol, solar_radius)

k_0_L = compute_luminosity(k_0_Fbol, k_star_radius)
k_m1_L = compute_luminosity(k_m1_Fbol, k_star_radius)
k_m2_L = compute_luminosity(k_m2_Fbol, k_star_radius)
k_m3_L = compute_luminosity(k_m3_Fbol, k_star_radius)

m_0_L = compute_luminosity(m_0_Fbol, m_star_radius)
m_m1_L = compute_luminosity(m_m1_Fbol, m_star_radius)
m_m2_L = compute_luminosity(m_m2_Fbol, m_star_radius)
m_m3_L = compute_luminosity(m_m3_Fbol, m_star_radius)
```

```python
print(f'G2V stellar lum.: {g_0_L}')
print(f'solar luminosity: {3.828e33}')

correction_ratio = 3.828e33 / g_0_L
print('correction ratio:', correction_ratio)
# correct the luminosities using this constant factor
f_0_L_corrected = f_0_L * correction_ratio
f_m1_L_corrected = f_m1_L * correction_ratio
f_m2_L_corrected = f_m2_L * correction_ratio
f_m3_L_corrected = f_m3_L * correction_ratio

g_0_L_corrected = g_0_L * correction_ratio
print('G0 corrected luminosity:', g_0_L_corrected)
g_m1_L_corrected = g_m1_L * correction_ratio
g_m2_L_corrected = g_m2_L * correction_ratio
g_m3_L_corrected = g_m3_L * correction_ratio

g_g1_L_corrected = g_g1_L * correction_ratio
g_g2_L_corrected = g_g2_L * correction_ratio
g_g3_L_corrected = g_g3_L * correction_ratio

k_0_L_corrected = k_0_L * correction_ratio
k_m1_L_corrected = k_m1_L * correction_ratio
k_m2_L_corrected = k_m2_L * correction_ratio
k_m3_L_corrected = k_m3_L * correction_ratio

m_0_L_corrected = m_0_L * correction_ratio
m_m1_L_corrected = m_m1_L * correction_ratio
m_m2_L_corrected = m_m2_L * correction_ratio
m_m3_L_corrected = m_m3_L * correction_ratio
```

```python
# compute flux received at Earth and Jupiter for these luminosities
def compute_flux(L, distance_cm):
    F = L / (4 * np.pi * distance_cm**2)
    return F
```

```python
g_0_F_Earth = compute_flux(g_0_L_corrected, Earth_distance_cm)
# compute fluxes at earth for the different metallicities
g_m1_F_Earth = compute_flux(g_m1_L_corrected, Earth_distance_cm)
g_m2_F_Earth = compute_flux(g_m2_L_corrected, Earth_distance_cm)
g_m3_F_Earth = compute_flux(g_m3_L_corrected, Earth_distance_cm)
# compute fluxes at earth for different surface gravities
g_g1_F_Earth = compute_flux(g_g1_L_corrected, Earth_distance_cm)
g_g2_F_Earth = compute_flux(g_g2_L_corrected, Earth_distance_cm)
g_g3_F_Earth = compute_flux(g_g3_L_corrected, Earth_distance_cm)
                            
# compute differences in flux between different metallicities
g_m1_diff = g_m1_F_Earth - g_0_F_Earth
g_m2_diff = g_m2_F_Earth - g_0_F_Earth
g_m3_diff = g_m3_F_Earth - g_0_F_Earth

g_g1_diff = g_g1_F_Earth - g_0_F_Earth
g_g2_diff = g_g2_F_Earth - g_0_F_Earth
g_g3_diff = g_g3_F_Earth - g_0_F_Earth

print(f'Total flux at Earth for G0: {g_0_F_Earth:.2e} erg/s/cm^2')
print(f'Flux metallicity differences: 0-1: {g_m1_diff:.2e}, 0-2: {g_m2_diff:.2e}, 0-3: {g_m3_diff:.2e}')
print(f'Flux surface gravity differences: 0-1: {g_g1_diff:.2e}, 0-2: {g_g2_diff:.2e}, 0-3: {g_g3_diff:.2e}')
```
